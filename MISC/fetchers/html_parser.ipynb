```python
#Not finished yet.

def html_parser(company_url_list):
    for companu_url in company_url_list:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
            "Accept-Language": "en-US,en;q=0.9",
            "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
            "Referer": "https://www.sec.gov/",
            "From": "azzagold99@gmail.com"  #email, remove this
        }

        r = requests.get(companu_url, headers=headers, timeout=10)
        r.raise_for_status()

        soup = BeautifulSoup(r.text, "html.parser")

        # Clean body text
        for tag in soup(["script", "style", "table"]):
            tag.extract()

        body_text = soup.body.get_text(separator="\n", strip=True)
        print(body_text[:8000])  # remove this after testing
        return body_text
```
